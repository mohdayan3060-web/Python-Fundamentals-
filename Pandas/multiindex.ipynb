Pivot table 

```python
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
new =df.groupby('sex')
new.total_bill.mean() 


df.groupby(['sex','smoker'])['total_bill'].mean().unstack()
```

```python
#using pviat table 
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='count')

#all colns togeather
df.pivot_table(index='sex',columns='smoker',aggfunc='mean',numeric_only=True)
```

```python
#multidimentional 
df.pivot_table(index=['sex','smoker'],columns=['day','time'],values='total_bill')
```

```python
#margins
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='mean', margins=True)
```

```python
#import csv file 
titanic=pd.read_csv(r'C:\Users\hp\Python-Fundamentals-\Pandas\titanic.csv')
titanic['Name']
```

```python
# comman function
titanic['Name'].str.upper()
titanic['Name'].str.capitalize()
titanic['Name'].str.len().max()
titanic['Name'][titanic['Name'].str.len()==82]
titanic['Name'].str.strip() #leading and tailing space remove karna hai 
```

```python
# split -> get 
#fetching lastname 
titanic['LastName']=titanic['Name'].str.split(',').str.get(0)
titanic
```

```python
#make new columns for name 
titanic[['tittle','firstname']]=titanic['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)

titanic['tittle'].value_counts()
```

```python
#replace 
titanic['tittle']=titanic['tittle'].str.replace('Ms.','Miss.')
titanic['tittle'].value_counts()
```

```python
#filtering 
#passangers name  startswith ('a')\endswith('A')
titanic[titanic['firstname'].str.startswith('A')]
titanic[titanic['firstname'].str.endswith('A')]

#isdigit
titanic[titanic['firstname'].str.isdigit()]
```

```python
#applying regular expressions (complex pattern fro searching )

titanic[titanic['firstname'].str.contains('john',case=False)]

#firstname end /start with aeiou
titanic[titanic['firstname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```

```python
#slicing 
titanic['Name'].str[::-1]
titanic['Name'].str[::2]
titanic['Name'].str[:4:2]


```
