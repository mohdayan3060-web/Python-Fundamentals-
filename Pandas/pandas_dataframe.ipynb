```python

#split apply combine staredgy 
#power of apply functions

# find the number movies starting with A for each group 
# Filter rows where Series_Title starts with 'A' within each group
def foo(group):
    return group["Series_Title"].str.startswith("A").sum()


# Apply the function to the grouped data
Genre.apply(foo)
```

```python
# find the rank of each group according to IMDB rating 
def rank_movie(group):
    group['genre_Rank']=group['IMDB_Rating'].rank(ascending=False)
    return group

Genre.apply(rank_movie).tail(20)
```

```python
#finding normalize rank
def normalize(group):
    group['norm_rating']=group['IMDB_Rating']-group['IMDB_Rating'].min()/group['IMDB_Rating'].max()-group['IMDB_Rating'].min()
    return group


Genre.apply(normalize).head(20)

```

```python
#redaing new csv file 
course=pd.read_csv(r'C:\Users\hp\Python-Fundamentals-\Pandas\courses.csv')
student02=pd.read_csv(r'C:\Users\hp\Python-Fundamentals-\Pandas\students.csv')
nov=pd.read_csv(r'C:\Users\hp\Python-Fundamentals-\Pandas\reg-month1.csv')
dec=pd.read_csv(r'C:\Users\hp\Python-Fundamentals-\Pandas\reg-month2.csv')
```

```python
student02
nov
dec
course
```

```python
# verticaly stack  concat
#no.colns make sure equall
regs=pd.concat([nov,dec],ignore_index=True) #index 
regs
```

```python
#multi index dataframe
multi=pd.concat([nov,dec],keys=['nov','dec'])
multi.loc[('nov',0)]
multi.loc[('dec',3)]
```

```python
#horizental stack 
pd.concat([nov,dec],axis=1)
```

```python
#merging
#inner joins(basis of comman colns)(only comman student will show )
# Inner Merge sirf wahi rows dikhata hai jo dono tables mein match karti hain

student02.merge(regs,how='inner' ,on='student_id')


```

```python
#left joing 
# Left Merge mein left side waali table ka saara data aayega. Right side waali table se sirf wahi data aayega jo left waale se match karega
course.merge(regs,how='left',on='course_id')
```

```python
#Right joining 
# . Right Merge mein right side waali table ka saara data aayega. Left table se sirf matching data uthaya jayega, aur jo match nahi karega wahan NaN aayega.
student02.merge(regs,how='right',on='student_id')
```

```python
#outer joins
# Outer Merge (Sab kuch le aao)
# Outer Merge (ya Full Outer Join) dono tables ka poora ka poora data combine kar deta hai. Jo match ho raha hai woh toh aayega hi, aur jo match nahi ho raha woh bhi aayega (missing values ki jagah NaN lag jayega).

student02.merge(regs,how='outer',on='student_id')
```
