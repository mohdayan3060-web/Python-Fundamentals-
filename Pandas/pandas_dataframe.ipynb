 GroupBy Object in Pandas

```python
deliveries=pd.read_csv(r'C:\Users\hp\Python fundamentals\Pandas\deliveries.csv')
imdb=pd.read_csv(r'C:\Users\hp\Python fundamentals\Pandas\imdb-top-1000 (1).csv')
```

```python
# Groupby is a powerful tool used to split a DataFrame into separate groups based on one or more columns
imdb

#applying buil-in aggregation function on groupby objects 
Genre=imdb.groupby('Genre')
Genre.sum()
Genre.min()
```

```python
#findout top 3 geners by total earning 
Genre.sum()['Gross'].sort_values(ascending=False).head(3)
# 2nd methods 
Genre['Gross'].sum().sort_values(ascending=False).head(3)
```

```python
# find out genres name have highest avg.Imdb reating

Genre.IMDB_Rating.mean().sort_values(ascending=False).head(1)
```

```python
#find out the direactorsname how get more votes (director popularity based on no_of_votes)
Director=imdb.groupby('Director')
Director.No_of_Votes.sum().sort_values(ascending=False).head(3)
```

```python
#find higest rated movie of each Gener
Genre=imdb.groupby('Genre')
x=Genre.IMDB_Rating.max().sort_values(ascending=False)


```

```python
#find out the no.of movies done by each actors 
Star1=imdb.groupby('Star1')
Star1.Series_Title.value_counts()
Star1.Series_Title.count()


```

GroupBy attributes and methods 

```python
#len 
len(Genre)

#value_count
Genre.size()

#find each group first item
Genre.first()

#find last movie each group

Genre.last()

#nth
Genre.nth(3)

#get group 
Genre.get_group('Fantasy') #fast and efficent to fetch particular group 

#groups
Genre.groups

#desccribe 
Genre.describe()

#sample
Genre.sample()

#nunique
Genre.nunique()

```

```python

#### Agg methods 
Genre.IMDB_Rating.mean()
##passing dict
Genre.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum'
    }
)
```
